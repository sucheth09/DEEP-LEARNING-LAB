Experiment 7 — Sigmoid Function

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z = np.arange(-5, 5, 0.1)
values = sigmoid(z)

plt.figure(figsize=(8, 5))
plt.plot(z, values, color="pink", linewidth=3)

plt.title("Visualization of the Sigmoid Function")
plt.xlabel("Input z")
plt.ylabel("Sigmoid(z)")
plt.grid(True)
plt.show()
```
